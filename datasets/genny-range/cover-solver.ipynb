```python
import csv
```

```python
sols = {}
with open('cover-solutions.txt', 'r', newline='') as f:
    reader = csv.reader(f, delimiter='\t')
    for row in reader:
        s = int(row[0])
        c = [int(i) for i in row[1].split(',')[:-1]]
        sols[s] = c
for s, c in sols.items():
    assert(len(c) == 1)
    sols[s] = c[0]
sols
```

```python
reverse_sols = {}
for s, c in sols.items():
    if c not in reverse_sols:
        reverse_sols[c] = []
    reverse_sols[c].append(s)
```

```python
wanted = [5, 100, 1000, 10000]
distances = {}
for c, s in reverse_sols.items():
    dist = []
    for w in wanted:
        dist.append(min(abs(i - w) for i in s))
    distances[c] = dist
distances
```
